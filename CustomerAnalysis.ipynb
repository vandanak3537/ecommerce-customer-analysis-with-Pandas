```python
import pandas as pd
import warnings
warnings.filterwarnings("ignore", category=FutureWarning)

df=pd.read_csv("data.csv")
```

```python
#creating a new column
df["Total_Amount"]=df["Quantity"]*df["Price"]
```

```python
print(df["Total_Amount"].sum())
```

```python
print(df["Total_Amount"].mean())
```

```python
print(df["Total_Amount"].max())
```

```python
print(df["Total_Amount"].min())
```

```python
#filtering
#spent more than 3000
print(df["Total_Amount"]>3000)
```

```python
#used UPI
print(df["Payment_Method"]=="UPI")
```

```python
#bought electronics
print(df["Category"]=="Electronics")
```

```python
#sorting
#find the 5 highest value orders
df.sort_values("Total_Amount",ascending=False).head(5)
```

```python
#group by
#revenue by category
df.groupby("Category")["Total_Amount"].sum()
```

```python
#revenue by city
df.groupby("City")["Total_Amount"].sum()
```

```python
#revenue by city
df.groupby("City")["Total_Amount"].sum()
```

```python
#revenue by payment methods
df.groupby("Payment_Method")["Total_Amount"].sum()
```

```python
#which category generated the most revenue
category_revenue=df.groupby("Category")["Total_Amount"].sum()
print(category_revenue)
print("Highest",category_revenue.idxmax())
```

```python
#which city has the highest spending
city_spend=df.groupby("City")["Total_Amount"].sum()
print(city_spend)
print("Highest: ",city_spend.idxmax())
```

```python
#which payment method is most popular
payment_count = df["Payment_Method"].value_counts()

print(payment_count)
```

```python
#which product sold the most unit
product_quantity=df.groupby("Product")["Quantity"].sum()
print(product_quantity)
print("Highest:", product_quantity.idxmax())
```

```python
#who spent the most
highest_spender=df.loc[df["Total_Amount"].idxmax()]
print(highest_spender["Name"])
print(highest_spender["Total_Amount"])
```

```python
#whats the average age of customers
print(df["Age"].mean())
```

```python
#which gender has higher spending
print(df.groupby("Gender")["Total_Amount"].sum())
```

```python
#which category has the highest average order value
print((df.groupby("Category")["Total_Amount"].mean()).idxmax)
```

```python
#find customers who spent more than 4k
high_spenders=df[df["Total_Amount"]>4000]
print(high_spenders[["Name","Total_Amount"]])
```

```python
#find the top 3 products bases on revenue
product_revenue=df.groupby("Product")["Total_Amount"].sum()
print(product_revenue.sort_values(ascending=False).head(3))
```
